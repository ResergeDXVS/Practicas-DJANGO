# Web Scraping de mercado libre
# Productos de documentos

```python
from selenium import webdriver
from bs4 import BeautifulSoup

driver = webdriver.Firefox()

def get_product_list():
    url = "https://www.amazon.com.mx/s?k=videojuegos&__mk_es_MX=%C3%85M%C3%85%C5%BD%C3%95%C3%91&crid=2G6NW05QFGCRM&sprefix=videojuegos%2Caps%2C190&ref=nb_sb_noss_1"
    driver.get(url)

    soup = BeautifulSoup(driver.page_source, "lxml")

    # Selecciona los títulos de productos en la lista
    a_tags = soup.find_all("a",class_="a-link-normal s-underline-text s-underline-link-text null s-link-style")
    print("-" * 50)
    for a in a_tags:
        link = a.get("href")
        detail_link = "https://www.amazon.com.mx" + link if "https" not in link else link
        
        print("\n")
        get_details(detail_link)

def get_details(url):
    driver.get(url)
    soup = BeautifulSoup(driver.page_source, "lxml")
    # Titulo
    title_tag = soup.find("span", id="productTitle")
    if title_tag:
        title = title_tag.get_text(strip=True)
    else:
        title = "Sin título"

    print("Titulo:", title)
    # Precio
    price_class = "a-offscreen"
    price_tag = soup.find("span", class_=price_class)
    if price_tag:
        price = price_tag.text.strip()
    else:
        price = "Sin precio"

    print("Precio:", price)

    # Descripción
    description_id = "featurebullets_feature_div"
    description_tag = soup.find("div", id=description_id)
    if description_tag:
        description = description_tag.text.strip()
    else:
        description = "Sin descripción"

    print("Descripción:", description)
    

    #Calificación
    tag = soup.find("span", class_="a-size-small a-color-base")

    if tag:
        print("Calificación:", tag.text)
    else:
        print("Sin calificación")

    print("-" * 50)
    
get_product_list()
driver.close()
```
